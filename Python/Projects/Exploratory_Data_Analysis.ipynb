```python
# Set the style for the plots
sns.set(style="whitegrid", context="notebook")

# Heatmap of missing values before removing columns
plt.figure(figsize=(12, 8))
sns.heatmap(df.isnull(), cbar=True, cmap='YlGnBu', linewidths=0.5)
plt.title('Heatmap of Missing Values', fontsize=16)
plt.xticks(rotation=90, fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.rcParams['figure.figsize']  = (20, 7)
```

```python
# Create a triangular heatmap of  correlation

# Mask the upper triangle of the correlation matrix
mask = np.triu(np.ones_like(df.corr(numeric_only=True), dtype=bool))

# Plotting the heatmap
plt.figure(figsize=(15, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='YlGnBu', linewidths=0.5, vmin=-1, vmax=1, mask=mask, fmt='.2f', cbar_kws={"shrink": .75}, annot_kws={"size": 7})
plt.title('Triangular Heatmap of Correlation of Numerical Variables', fontsize=16)
plt.xticks(rotation=90, fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```
