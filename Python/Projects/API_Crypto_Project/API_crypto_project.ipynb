```python
# Now let's look at the coin trends over time

df3 = df.groupby('name', sort=False)[['quote.USD.percent_change_1h','quote.USD.percent_change_24h','quote.USD.percent_change_7d','quote.USD.percent_change_30d','quote.USD.percent_change_60d','quote.USD.percent_change_90d']].mean()
df3
```

```python
df4 = df3.stack()
df4
```

```python
df10 = df[['name','quote.USD.price','timestamp']]
df10 = df10.query("name == 'Bitcoin'")
df10
```

```python
df10 = df[['name','quote.USD.price','timestamp']].query("name == 'Bitcoin'").copy()
df10['timestamp'] = pd.to_datetime(df10['timestamp'])
df10 = df10.sort_values('timestamp')

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df10['timestamp'], df10['quote.USD.price'])

# mniej etykiet + ładny format
ax.xaxis.set_major_locator(mdates.AutoDateLocator())
ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(ax.xaxis.get_major_locator()))

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
